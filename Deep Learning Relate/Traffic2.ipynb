```python
import cv2
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture("traffic.mp4")
vehicle_classes = [2]

while True:
    ret, frame = cap.read()

    if not ret:
        break
    results = model(frame)
    count = 0
    vehicles = []

    for result in results:
        boxes = result.boxes
        for box in boxes:
            cls = int(box.cls[0])
            if cls in vehicle_classes:
                count += 1
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                vehicles.append((x1, y1, x2, y2))
    if count <= 5:
        signal = "GREEN"
        active = "green"
        color = (0,255,0)
    elif count <= 10:
        signal = "YELLOW"
        active = "yellow"
        color = (0,255,255)
    else:
        signal = "RED"
        active = "red"
        color = (0,0,255)
    cv2.putText(frame,
                "Vehicles: " + str(count),
                (20,40),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                color,
                2)
    cv2.putText(frame,
                "Signal: " + signal,
                (20,80),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                color,
                2)
    cv2.rectangle(frame,
                  (500,20),
                  (560,170),
                  (50,50,50),
                  -1)
    cv2.circle(frame,
               (530,50),
               15,
               (0,0,255),
               -1)
    cv2.circle(frame,
               (530,95),
               15,
               (0,255,255),
               -1)
    cv2.circle(frame,
               (530,140),
               15,
               (0,255,0),
               -1)
    for (x1,y1,x2,y2) in vehicles:
        cv2.rectangle(frame,
                      (x1,y1),
                      (x2,y2),
                      (0,255,0),
                      2)
    cv2.imshow("YOLO Traffic Detection", frame)
    if cv2.waitKey(30) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
